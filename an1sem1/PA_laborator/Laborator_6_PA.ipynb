**Sortarea unui vector după un criteriu — parametrul `key`**


### SORTĂRI LISTE — a)
Se consideră o listă de numere. Să se sorteze folosind o funcție `lambda` așa cum s-ar fi sortat dacă numerele erau șiruri.

````python
nr = [6 , 3 , 12, 4, 2, 1112, 113]

fn = lambda x: str(x)

print(sorted(nr, key=fn))
````

### SORTĂRI LISTE — b)
Se consideră o listă de numere. Să se sorteze folosind o funcție `lambda` comparând întâi ultima cifră, apoi penultima etc. (ca și cum ar fi șiruri cu literele în ordine inversă).

````python
nr = [6 , 3 , 12, 4, 2, 1112, 113]

fn = lambda x: str(x)[::-1]

print(sorted(nr, key=fn))
````

### SORTĂRI LISTE — c)
Se consideră o listă de numere. Să se sorteze descrescător după **lungimea** numărului.

````python
nr = [6 , 3 , 12, 4, 2, 1112, 113]

fn = lambda x: len(str(x))

print(sorted(nr, key=fn))
````

### SORTĂRI LISTE — d)
Se consideră o listă de numere. Să se sorteze după **numărul de cifre distincte**.

````python
nr = [6 , 3 , 12, 4, 2, 1112, 113]

fn = lambda x: len(set(str(x)))

print(sorted(nr, key=fn))
````

### SORTĂRI LISTE — e)
Sortare numere **întâi după lungime** și apoi **descrescător după valoare**.

````python
nr = [6 , 3 , 12, 4, 2, 1112, 113]

fn = lambda x: (str(x), -x)

print(sorted(nr, key=fn))
````

### TEMA — f)
Să se sorteze o listă de numere naturale astfel încât **numerele impare sortate crescător** să fie urmate de **numerele pare sortate descrescător**.

````python
nr = [6 , 3 , 12, 4, 2, 1112, 113]

fn = lambda x: (x % 2 == 0, x if x % 2 != 0 else -x)

print(sorted(nr, key=fn))
````

### TEMA — g)
Se citește o propoziție cu cuvinte separate prin spațiu. Să se formeze o nouă propoziție cu cuvintele din prima propoziție care au **lungime cel puțin 2** ordonate **descrescător** după lungime.

````python
words = "Ana are mere dulci si bune rau de tot".split()

good_words = filter(lambda word: (len(word) >= 2), words)
fn = lambda word: len(word)

print(*sorted(good_words, key = fn, reverse = True))
````

### TEMA — h)
Se citește un vector de numere naturale (o linie, separate prin spațiu). Să se ordoneze elementele **crescător după suma cifrelor**, iar în caz de egalitate, **descrescător după valoare**.

Exemplu:
```
v = [11, 45, 20, 810, 179, 81, 1000]
=> v = [1000, 20, 11, 810, 81, 45, 179]
```

````python
def sum_cif(x):
    x = str(x)
    return sum(int(ch) for ch in x)

nr = [6 , 3 , 12, 4, 2, 1112, 113]

fn = lambda x: (sum_cif(x), -x)

print(sorted(nr, key=fn))
````

**f)** Scrieți o funcție care primește ca parametru structura de date (de la a) și **adaugă** la informațiile fiecărui student **un cod** de lungime 6 generat aleator care conține **3 litere** urmate de **3 cifre**.
Exemplu de apel:
```python
genereaza_coduri(d)
print(d)
```


````python


def get_elevi(path):
    elevi = {}
    with open(path, "r") as f:
        lines = f.readlines()
        for line in lines:
            line = line.strip().split()
            if len(line) < 3:
                continue
            cnp = line[0]
            nume = line[1]
            prenume = line[2]
            note = []
            for elem in line[3:]:
                try:
                    if int(elem) <= 10:
                        note.append(int(elem))
                except ValueError:
                    pass
            
            elevi[cnp] = {
                "nume": nume,
                "prenume": prenume,
                "note": note
            }
    return elevi

dict = get_elevi("elevi.in")
print(dict)
````

### Procesare fișiere cu frecvențe de cuvinte și distanță cosinus
**a)** Scrieți o funcție care primește ca parametru **două nume de fișiere** (variantă: un număr variabil de nume) și **returnează un dicționar** cu cuvintele care apar în **cel puțin unul** dintre fișiere și **frecvența totală** cu care apare fiecare cuvânt (suma frecvențelor din fișiere). Cuvintele pot fi pe mai multe linii; pe o linie sunt separate prin spații.

**b)** Pentru fișierele `cuvinte1.in` și `cuvinte2.in`, **afișați cuvintele** care apar în cel puțin unul dintre fișiere **ordonate crescător lexicografic**.

**c)** Pentru `cuvinte1.in`, **creați o listă de perechi** `(cuvânt, frecvență)` cu cuvintele și frecvența lor, **ordonată descrescător după frecvență** (folosind funcția de la a)).

**d)** Determinați un cuvânt care apare **cel mai des** în `cuvinte2.in`, folosind funcția de la a) și `max`. Dacă sunt mai multe posibilități, afișați **cuvântul cel mai mic lexicografic**.

**e)** Distanța **cosinus** dintre două documente `F1` și `F2`. Fie `{c1, c2, …, cn}` mulțimea cuvintelor care apar în cel puțin unul. Pentru `1≤i≤n`, `v_i1`, `v_i2` sunt numerele de apariții ale cuvântului `i` în primul, respectiv al doilea document. Formula similarității cosinus dintre doi vectori **F₁** și **F₂** este definită astfel:

$$
\text{dcos}(F_1, F_2) = \frac{\sum_{i=1}^{n} v_{i1} \cdot v_{i2}}{\sqrt{\sum_{i=1}^{n} v_{i1}^2} \cdot \sqrt{\sum_{i=1}^{n} v_{i2}^2}}
$$

Unde:
- $v_{i1}$ este componenta *i* a vectorului **F₁**
- $v_{i2}$ este componenta *i* a vectorului **F₂**

Să se calculeze **cu două zecimale** distanța cosinus dintre fișierele `caractere1.in` și `caractere2.in` (folosind funcția de la a) apelată separat pentru fiecare fișier).

**Exemplu** (rezumat din cerință) pentru fișierele:
<pre>
a fost 
a fost
odata ca in povesti
o data ca in povesti
in povesti
</pre>
și 
<pre>
a fost 
a fost
altadata ca in basme
alta data ca in basme
nu in povesti
</pre>
- Dicționarul va fi: `{'a': 4, 'fost': 4, 'odata': 1, 'ca': 4, 'in': 6, 'povesti': 4, 'o': 1, 'data': 2, 'altadata': 1, 'basme': 2, 'alta': 1, 'nu': 1}`
- Cuvinte ordonate: `a alta altadata basme ca data fost in nu o odata povesti`
- Perechi (descrescător după frecvență): `[('in', 3), ('povesti', 3), ('a', 2), ('fost', 2), ('ca', 2), ('odata', 1), ('o', 1), ('data', 1)]`
- Cuvânt maxim în `cuvinte2.in`: `in`
- dcos ≈ `0.79`
