### Exercițiul 1
Să se copieze conținutul unui fișier text în alt fișier folosind diferite funcții de citire din fișier (`read`, `readline`, `readlines`, iterarea liniilor cu `for`).

````python
# READ

f = open("input.txt", "r")
content = f.read()

print(content)
f.close()

g = open("output.txt", "w")
g.write(content)
g.close()
````

````python
with open("file.txt", "r"):
    content = f.read()
    print(content)

with open("out.txt", "w"):
    g.write(content)
````

````python
# READLINE

with open("input.txt", "r") as f, open("output.txt", "w") as g:
    line = f.readline()
    while line:
        g.write(line)
        line = f.readline()
````

````python
# READLINES

with open("input.txt", "r") as f, open("output.txt", "w") as g:
    lines = f.readlines()
    for line in lines:
        g.write(line)
````

````python
# FOR -> sub forma de string

with open("input.txt", "r") as f, open("output.txt", "w") as g:
    for line in f:
        print(type(line))
        g.write(line)
````

### Exercițiul 2
În fișierul text `test.in` se află testul unui elev de clasa a II-a la matematică, conținând 9 înmulțiri scrise pe rânduri distincte. Un calcul corect este notat cu un punct, iar unul incorect cu 0 puncte. Să se realizeze un program care să evalueze testul dat, astfel: în dreptul fiecărui calcul corect se va scrie mesajul `Corect`, iar în dreptul fiecărui calcul greșit se va scrie mesajul `Gresit` și rezultatul corect, iar la final se va scrie nota (un punct se acordă din oficiu!). Rezultatul evaluării testului se va afișa în fișierul `test.out`.

**test.in**
```
3*4=11
2*10=20
5*5=24
7*4=28
2*6=12
10*10=100
3*9=27
6*7=33
0*9=1
```

**test.out**
```
3*4=11 Gresit 12
2*10=20 Corect
5*5=24 Gresit 25
7*4=28 Corect
2*6=12 Corect
10*10=100 Corect
3*9=27 Corect
6*7=33 Gresit 42
0*9=1 Gresit 0
Nota 6
```

````python
nota = 1

with open("input.txt", "r") as f, open("output.txt", "w") as g:
    equations = f.readlines()
    for eq in equations:
        prompt = eq.strip()
        eq = eq.split("=")
        oper = eq[0].split("*")
        result = int(eq[1])
        a = int(oper[0])
        b = int(oper[1])

        if a * b == result:
            g.write(f"{prompt} Corect\n")
            nota += 1
        else:
            g.write(f"{prompt} Gresit {a * b}\n")
    g.write(f"Nota {nota}")
````

### Exercițiul 3
Se citesc m, n și o matrice cu m linii și n coloane (elementele unei linii separate cu spațiu). Să se creeze o listă cu maximele de pe fiecare linie (folosind și comprehensiune).

````python
with open("input.txt", "r") as f, open("output.txt", "w") as g:
    m, n = map(int, f.readline().split())
    matrice = [list(map(int, f.readline().split())) for _ in range(m)]
    maxime = [max(linie) for linie in matrice]
    print(matrice)
    print(maxime)
````

### Exercițiul 4
Se citesc m, n și o matrice cu m linii și n coloane (elementele unei linii separate cu spațiu). Se citește în plus un număr natural k. Să se insereze o linie nouă cu toate elementele 0 între liniile k și k+1 ale matricei.

````python
with open("input.txt", "r") as f, open("output.txt", "w") as g:
    m, n, k = map(int, f.readline().split())
    matrice = [list(map(int, f.readline().split())) for _ in range(m)]
    matrice.insert(k - 1, [0] * n)
    print(matrice)

    for line in matrice:
        g.write(" ".join(map(str, line)) + "\n")
````

### Exercițiul 5
Se consideră tabloul bidimensional cu m linii și n coloane, care conține doar valorile {0,1,2}. Să se determine câte linii au produsul elementelor maxim (folosind și comprehensiune).

````python
with open("input.txt", "r") as f, open("output.txt", "w") as g:
    m, n = map(int, f.readline().split())
    matrice = [list(map(int, f.readline().split())) for _ in range(m)]
    produs = max([line.count(2) for line in matrice if 0 not in line])
    count  = sum([1 for line in matrice if line.count(2) == produs and 0 not in line])
    print(count)
    for line in matrice:
        g.write(" ".join(map(str, line)) + "\n")
````
