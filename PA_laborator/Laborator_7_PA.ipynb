### FUNCȚII
a) Scrieți o funcție care să citească de la tastatură o listă cu elemente numere întregi. Numărul de elemente ale listei și elementele sale se vor citi în cadrul funcției.

b) Scrieți o funcție care primește ca parametru o secvență `s`, un element `x` și, opțional, doi indici `i` și `j` și returnează poziția **primului element mai mare decât `x`** din `s[i:j]` (dacă `i` sau `j` nu se specifică, atunci comportamentul va fi cel de la feliere) și `-1` în caz că nu există un astfel de element.

c) Scrieți un program care, folosind apeluri utile ale funcției definite anterior, afișează mesajul **"Da"** în cazul în care o **listă de numere întregi**, citită de la tastatură, este **sortată strict descrescător** sau mesajul **"Nu"** în caz contrar. **Aceeași cerință** și pentru **o listă de cuvinte**.


````python

def read(n, lst):
    n = int(input().strip())
    for i in range(n):
        lst.append(input().strip())
    return n, lst


def over_x(secv, x):
    for el in secv:
        if el > x:
            return el
    return -1

def is_sorted(secv):
    for i in range(1, len(secv)):
        if over_x(secv[i::], secv[i]) != -1:
            return False
    return True


def main():
    n = 0
    lst = []
    read(n, lst)
    if is_sorted(lst):
        print("YES")
    else:
        print("NO")


if __name__ == "__main__":
    main()
````

### Funcții cu număr variabil de parametri (I)
a) Scrieți o funcție cu **număr variabil de parametri** care să furnizeze **numărul natural obținut prin alipirea cifrelor maxime** ale numerelor naturale nenule primite ca parametri.
Exemplu: pentru numerele `4251, 73, 8, 133` funcția → `5783`.

b) Scrieți o funcție cu **3 parametri nenuli** de tip întreg `a, b, c` care să verifice dacă aceștia pot fi considerați ca fiind **numere scrise în baza 2** sau nu, folosind apeluri utile ale funcției definite anterior.
Exemple: `(1001, 11, 100)` → `True`; `(1001, 17, 100)` → `False`.


````python

def cif_max(x):
    lst = str(x)
    lst = lst.replace('', ' ').split()
    return max(lst)

def make_number(*argv):
    number = ''
    for arg in argv:
        number += str(cif_max(arg))
    return number

def numbers_3(a, b, c):
    result = make_number(a, b, c)
    result = list(map(int, result.replace('', ' ').split()))
    if max(result) > 1:
        return False
    return True

print(numbers_3(101, 1111101100, 110))
````

````python

def search_word_in_file(word, path):
    indexes = []
    with open(path, 'r') as f:
        index = 1
        for line in f:
            if word in line:
                indexes.append(index)
            index += 1
    return indexes

def files(word, *paths):
    outputs = []
    for path in paths:
        outputs.append(search_word_in_file(word, path))
    return outputs

print(files("buna", "file1.txt", "file2.txt"))
````

### (Model) `cinema.in` — structură de date și interogări
Fișierul text `cinema.in` conține programul dintr-o zi al unui lanț de cinematografe. Fiecare linie:
```
nume_cinematograf % nume_film % ore_de_difuzare
```
- `nume_cinematograf`: șir (poate conține spații, fără caracterul `%`)
- `nume_film`: șir (poate conține spații, fără caracterul `%`)
- `ore_de_difuzare`: ore `hh:mm` separate prin spațiu

**Exemplu**:
```
Cinema 1 % Minionii 2 % 12:30 18:30
Cinema 3 % Elfii cofetari % 10:30 12:30
Cinema 2 % Minionii 2 % 15:00 18:30 20:30
Cinema 1 % Elfii cofetari % 10:00 12:30
Cinema 2 % Gasca Animalutelor % 15:00 18:30 20:00
Cinema 4 % Minionii 2 % 16:00 18:30 20:30
Cinema 1 % Buna dimineata % 09:30
```

a) **[2,5 p.]** Memorați datele din fișier **într-o singură structură** de date eficientă pentru cerințele de mai jos.

b) **[1 p.]** Funcția `sterge_ore(struct, cinema, film, ore: set[str])` — șterge din programările cinematografului `cinema` orele din `ore` pentru filmul `film` și **returnează o listă** cu filmele programate la `cinema` **după actualizare**. Citiți de la tastatură `f` (film), `c` (cinema) și `o` (ora `hh:mm`), apelați funcția și afișați **lista returnată**; apoi afișați **structura** actualizată.

c) **[1,5 p.]** Funcția `cinema_film(struct, *cinematografe, ora_minima, ora_maxima)` → **listă de tupluri** `(nume_film, nume_cinema, lista_de_ore)` cu filmele care încep la cel puțin unul dintre `cinematografe` **între** `ora_minima` și `ora_maxima`. `lista_de_ore` se sortează crescător. **Lista returnată** se ordonează **crescător după numele filmului** și, la egalitate, **descrescător** după numărul elementelor din `lista_de_ore`.
Apel de test: cinematografe `('Cinema 1','Cinema 2')`, `ora_minima="14:00"`, `ora_maxima="22:00"`.
Exemplu rezultat (pentru datele de mai sus):
```
[('Gasca Animalutelor', 'Cinema 2', ['15:00', '18:30', '20:00']),
 ('Minionii 2', 'Cinema 2', ['15:00', '18:30', '20:30']),
 ('Minionii 2', 'Cinema 1', ['18:30'])]
```
(Notă: ‘Elfii cofetari’ nu apare — programat înainte de 14:00.)


````python
dict = {}
file_path = "cinema.in"

def decode_line(line):
    parts = line.split("%")
    hours = parts[2].split()
    return parts[0], parts[1], hours

def read_file():
    with open(file_path, "r") as f:
        for line in f:
            cinema_name, movie, hours = decode_line(line)
            if dict.get(cinema_name) is None:
                lst = []
                lst.append([movie, hours])
                dict[cinema_name] = {"movies": lst}
            else:
                dict[cinema_name]["movies"].append([movie, hours])


def main():
    read_file()
    for cinema in dict:
        print(f"Cinema: {cinema}")
        for movie in dict[cinema]["movies"]:
            print(f" Movie: {movie[0]}")
            print(f"  Hours: {movie[1]}")


if __name__ == "__main__":
    main()
````
