## 1. Recap: Functii simple

O functie in Python se defineste cu `def`:
```python
def salut(nume):
    print(f"Salut, {nume}!")
```
Apoi o putem apela:
```python
salut("Ana")
```


````python
# Exemplu de functie simpla
def salut(nume):
    print(f"Salut, {nume}!")

salut("Ana")
salut("Mihai")
````

## 2. Parametri default

Putem sa oferim valori implicite (default) parametrilor. Daca nu trimitem o valoare cand apelam functia, valoarea default este folosita.

```python
def salut(nume, mesaj="Bine ai venit"):
    print(f"{mesaj}, {nume}!")
```

- Daca apelam `salut("Ana")` -> mesajul va fi "Bine ai venit".
- Daca apelam `salut("Ana", "Ce mai faci")` -> mesajul va fi "Ce mai faci".


````python
# Exemplu: parametri default
def salut(nume, mesaj="Bine ai venit"):
    print(f"{mesaj}, {nume}!")

salut("Ana")           # Foloseste mesajul default
salut("Mihai", "Salutari")  # Suprascrie mesajul default
````

### 2.1 Reguli de ordonare a parametrilor

In Python, ordinea parametrilor intr-o functie conteaza:
1. Parametri pozitionali simpli (fara valoare default)
2. Parametri cu valoare default
3. `*args` (argumente pozitionale variabile)
4. Parametri keyword-only (optional)
5. `**kwargs` (argumente cheie=valoare variabile)

De exemplu:
```python
def exemplu(a, b=10, *args, c=20, **kwargs):
    pass
```


````python
def exemplu(a, b=10, *args, c=20, **kwargs):
    print("a =", a)
    print("b =", b)
    print("args =", args)
    print("c =", c)
    print("kwargs =", kwargs)
    print("-" * 40)


# 1. Doar argumentul obligatoriu
exemplu(1)
# a = 1
# b = 10 (implicit)
# args = ()
# c = 20 (implicit)
# kwargs = {}

# 2. Suprascrierea lui b
exemplu(1, 2)
# a = 1
# b = 2
# args = ()
# c = 20
# kwargs = {}

# 3. Folosirea argumentelor variabile (*args)
exemplu(1, 2, 3, 4, 5)
# a = 1
# b = 2
# args = (3, 4, 5)
# c = 20
# kwargs = {}

# 4. Setarea lui c (numai prin argument numit!)
exemplu(1, c=99)
# a = 1
# b = 10
# args = ()
# c = 99
# kwargs = {}

# 5. *args + argument numit pentru c
exemplu(1, 2, 3, 4, c=50)
# a = 1
# b = 2
# args = (3, 4)
# c = 50
# kwargs = {}

# 6. Argumente suplimentare numite (**kwargs)
exemplu(1, x=100, y=200)
# a = 1
# b = 10
# args = ()
# c = 20
# kwargs = {'x': 100, 'y': 200}

# 7. Combinatie completa
exemplu(1, 2, 3, 4, 5, c=88, x=10, y=20)
# a = 1
# b = 2
# args = (3, 4, 5)
# c = 88
# kwargs = {'x': 10, 'y': 20}

# 8. Setarea lui b prin argument numit (nu ai avut acest exemplu)
exemplu(1, b=99)
# a = 1
# b = 99
# args = ()
# c = 20
# kwargs = {}

# 9. Setarea lui c + kwargs, fara *args
exemplu(1, c=33, x=1, y=2)
# a = 1
# b = 10
# args = ()
# c = 33
# kwargs = {'x': 1, 'y': 2}

# 10. Doar kwargs suplimentare (pe langa argumentul obligatoriu)
exemplu(1, k1=10, k2=20)
# a = 1
# b = 10
# args = ()
# c = 20
# kwargs = {'k1': 10, 'k2': 20}

# 11. *args, dar fara b (folosind b implicit)
exemplu(1, *[10, 20, 30])
# a = 1
# b = 10
# args = (10, 20, 30)
# c = 20
# kwargs = {}

# 12. b numit + c numit + kwargs
exemplu(1, b=5, c=7, extra=True)
# a = 1
# b = 5
# args = ()
# c = 7
# kwargs = {'extra': True}

# 13. *args si **kwargs fara niciun argument numit special
exemplu(1, 2, 3, val=100)
# a = 1
# b = 2
# args = (3,)
# c = 20
# kwargs = {'val': 100}

# 14. Toate tipurile, dar in alt aranjament decat cele prezentate
exemplu(1, 2, *("a", "b"), c="X", mode="test", debug=False)
# a = 1
# b = 2
# args = ('a', 'b')
# c = 'X'
# kwargs = {'mode': 'test', 'debug': False}

# 15. Numai argumentul a si c, restul toate implicite
exemplu(1, c=123)
# a = 1
# b = 10
# args = ()
# c = 123
# kwargs = {}

# 16. b pozitional + c numit + multe kwargs
exemplu(1, 22, c=33, x=1, y=2, z=3)
# a = 1
# b = 22
# args = ()
# c = 33
# kwargs = {'x': 1, 'y': 2, 'z': 3}
````

## 3. `*args` - numar variabil de argumente pozitionale

`*args` iti permite sa trimiti orice numar de argumente pozitionale catre o functie. In interiorul functiei, `args` este un tuple.

```python
def suma(*args):
    # args este un tuple cu toate argumentele pozitionale primite
    total = 0
    for x in args:
        total += x
    return total
```


````python
# Exemplu: *args pentru a calcula suma unor numere
def suma(*args):
    total = 0
    for x in args:
        total += x
    return total

print(suma(1, 2))
print(suma(1, 2, 3, 4))
print(suma())  # fara argumente
````

## 4. `**kwargs` - numar variabil de argumente cheie=valoare

`**kwargs` iti permite sa trimiti orice numar de argumente cheie=valoare catre o functie. In interiorul functiei, `kwargs` este un dict.

```python
def afiseaza_info(**kwargs):
    for cheie, valoare in kwargs.items():
        print(cheie, "->", valoare)
```


````python
# Exemplu: **kwargs pentru a afisa informatii flexibile
def afiseaza_info(**kwargs):
    for cheie, valoare in kwargs.items():
        print(f"{cheie} -> {valoare}")

afiseaza_info(nume="Ana", varsta=25, oras="Cluj")
afiseaza_info(titlu="Python", nivel="incepator")
````

## 5. Combinarea parametrilor simpli, default, `*args` si `**kwargs`

Putem combina mai multe tipuri de parametri intr-o singura functie. Important este sa respectam ordinea corecta.

```python
def exemplu(mesaj, prefix="[INFO]", *args, sep=", ", **kwargs):
    # mesaj  - parametru pozitional obligatoriu
    # prefix - parametru cu valoare default
    # *args  - argumente pozitionale suplimentare
    # sep    - parametru keyword-only cu valoare default
    # **kwargs - argumente cheie=valoare suplimentare
    pass
```


````python
def exemplu(mesaj, prefix="[INFO]", *args, sep=", ", **kwargs):
    # Afisam argumentele primite pentru vizualizare
    print("mesaj =", mesaj)
    print("prefix =", prefix)
    print("args =", args)
    print("sep =", sep)
    print("kwargs =", kwargs)
    print("-" * 40)


# 1. Doar parametrul obligatoriu
exemplu("Salut")
# mesaj = "Salut"
# prefix = "[INFO]"
# args = ()
# sep = ", "
# kwargs = {}

# 2. Suprascrierea lui prefix
exemplu("Pornit", "[DEBUG]")
# prefix = "[DEBUG]"

# 3. Un singur argument suplimentar (*args)
exemplu("Mesaj", "[INFO]", 10)
# args = (10,)

# 4. Mai multe argumente pozitionale (*args)
exemplu("Start", "[SYS]", 1, 2, 3)
# args = (1, 2, 3)

# 5. Folosirea parametrului keyword-only sep
exemplu("Lista", "[OK]", 1, 2, 3, sep=" | ")
# sep = " | "

# 6. Trimitem argumente suplimentare numite (**kwargs)
exemplu("Login", user="admin", success=True)
# kwargs = {"user": "admin", "success": True}

# 7. prefix + **kwargs
exemplu("Logout", "[WARN]", reason="timeout")
# prefix = "[WARN]"
# kwargs = {"reason": "timeout"}

# 8. Combinatie completa: prefix + *args + sep + **kwargs
exemplu("Proces", "[DBG]", 10, 20, 30, sep=" - ", mode="auto", retry=3)
# args = (10, 20, 30)
# sep = " - "
# kwargs = {"mode": "auto", "retry": 3}

# 9. Fara prefix (default) + *args + **kwargs
exemplu("Calculez", 1, 2, 3, metoda="fast")
# prefix = "[INFO]"
# args = (1, 2, 3)
# kwargs = {"metoda": "fast"}

# 10. Apel cu prefix numit
exemplu("Mesaj important", prefix="[CRITICAL]")
# prefix = "[CRITICAL]"

# 11. Apel cu prefix si sep numite
exemplu("Export", prefix="[SYS]", sep=" :: ")
# prefix = "[SYS]"
# sep = " :: "

# 12. args expandat dintr-un iterable
lista = [100, 200, 300]
exemplu("Valori", "[DATA]", *lista)
# args = (100, 200, 300)
````

````python
# Exemplu combinat
def log_mesaj(mesaj, prefix="[INFO]", *args, sep=", ", **kwargs):
    """Afiseaza un mesaj de log cu optiuni flexibile."""
    linie = f"{prefix} {mesaj}"
    if args:
        linie += " " + sep.join(str(a) for a in args)
    if kwargs:
        extra = "; ".join(f"{k}={v}" for k, v in kwargs.items())
        linie += f" | {extra}"
    print(linie)

log_mesaj("Pornire aplicatie")
log_mesaj("Valori curente", 1, 2, 3)
log_mesaj("Conectare utilizator", user="ana", rol="admin")
log_mesaj("Debug", 10, 20, mod="test", sep=" - ")
````

## 6. Despachetarea argumentelor cu `*` si `**`

Daca avem deja o lista / tuplu sau un dictionar, le putem despacheta in argumente:

```python
def aduna(a, b, c):
    return a + b + c

valori = (1, 2, 3)
rez = aduna(*valori)  # echivalent cu aduna(1, 2, 3)
```

Pentru `**`, despachetam un dictionar:
```python
param = {"a": 1, "b": 2, "c": 3}
rez = aduna(**param)
```


````python
# Exemplu: despachetarea argumentelor
def aduna(a, b, c):
    return a + b + c

valori = (1, 2, 3)
print(aduna(*valori))

param = {"a": 4, "b": 5, "c": 6}
print(aduna(**param))
````

## 8. Exemple avansate (doar pentru studiu)

In aceasta sectiune sunt cateva exemple mai avansate care folosesc `*args` si `**kwargs` in moduri utile:
- un decorator `debug` care afiseaza numele functiei si argumentele primite,
- o functie `merge_dicts` care combina mai multe dictionare folosind `*args` si `**kwargs`.


````python
# Decorator avansat de debug
from functools import wraps

def debug(func):
    """Decorator care afiseaza numele functiei si argumentele la fiecare apel."""
    @wraps(func)
    def wrapper(*args, **kwargs):
        print(f"[DEBUG] Apel {func.__name__} cu args={args}, kwargs={kwargs}")
        rezultat = func(*args, **kwargs)
        print(f"[DEBUG] {func.__name__} a intors {rezultat}")
        return rezultat
    return wrapper

@debug
def calculeaza(a, b):
    return 2 * a + b

print(calculeaza(3, 7))
print(calculeaza(2, 5))
print(calculeaza(b=3, a=7))
print(calculeaza(b=2, a=5))
````

````python
# Exemplu: combinarea mai multor dictionare
def merge_dicts(*args, **kwargs):
    """Combina mai multe dictionare intr-unul singur.
    Dictionarele din kwargs au prioritate mai mare.
    """
    rezultat = {}
    for d in args:
        rezultat.update(d)
    rezultat.update(kwargs)
    return rezultat

d1 = {"a": 1, "b": 2}
d2 = {"b": 3, "c": 4}
print(merge_dicts(d1, d2, d=5, e=6))
````

````python
#Prioritate mai mare
d1 = {"a": 1, "b": 2}
d2 = {"b": 20, "c": 3}

rezultat = merge_dicts(d1, d2, b=200, d=4)
print(rezultat)
````
