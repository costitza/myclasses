**Exercițiu:** Se citește o propoziție. Să se înlocuiască cifrele mai mici decât 5 cu denumirea lor ('1' → 'unu', etc.).

```python
import inflect 

def checkInteger(s):
      
    if s.isdigit():
        return True
    return False

text = input().strip().split()

p = inflect.engine()

new_text = []
for elem in text:
    if checkInteger(elem):
        if 0 <= int(elem) <= 4:
            new_text.append(p.number_to_words(int(elem)))
        else:
            new_text.append(elem)
    else:
        new_text.append(elem)

print(" ".join(new_text))
```

**Exercițiu:** Împărțiți un text după multiple tipuri de separatori (ex: orice caracter care nu este literă sau cifră).

```python
import re

text = "Ana_are-32.de*mere! și? pere12"
# Împărțim după orice caracter care NU este literă sau cifră
rezultat = re.split(r'[^A-Za-z0-9]+', text)

print(rezultat)
```

**Exercițiu:** Înlocuiți toate aparițiile unui cuvânt cu altul, doar dacă este precedat de un anumit șir.

```python
import re 

text = "cam mic, destul mic, foarte mic"

to_change = "mic"
new = "mare"
prefix = "cam"

pattern = rf"{prefix} {to_change}"

new_text = re.sub(pattern, prefix + " " + new, text)
print(new_text)
```

**Exercițiu:** Extrageți toate numerele dintr-un text.

**Exercițiu:** Extrageți toate adresele IP dintr-un text.

```python
import re 

pattern = r"\b(?:(?:25[0-5]|2[0-4]\d|1\d{2}|[1-9]?\d)\.){3}(?:25[0-5]|2[0-4]\d|1\d{2}|[1-9]?\d)\b"

text = """
Serverul principal răspunde la 192.168.1.10, iar serverul secundar la 10.0.0.254.
Am observat trafic suspect de la 255.255.255.255 și de la 172.16.300.1 (care pare invalid).
Mai există un dispozitiv vechi la 192.168.0.1 și unul temporar la 8.8.8.8 folosit ca DNS.
Uneori mai apare și adresa 127.0.0.1 în loguri. Pe un laptop a fost găsită și 999.12.3.4 care clar nu este validă.

"""

ips = re.findall(pattern, text)

print(*ips)
```

**Exercițiu:** Extrageți toate atributele HTML dintr-un tag (hint: regex + Hackerrank).

```python
import re

text = '<a href="https://www.hackerrank.com" class="nav-link" data-id=123 disabled>'

pattern = r'([a-zA-Z_:][a-zA-Z0-9:._-]*)\s*(?:=\s*(".*?"|\'.*?\'|[^\s>]+))?'

tags = re.findall(pattern, text)

atributes = [name for name, value in tags]

print(*atributes)
```

**Exercițiu:** Detectați toate domeniile dintr-un text (ex: google.com, openai.org).

```python
import re

text = "Vizitează google.com, mail.google.com și openai.org pentru detalii."
pattern = r"\b(?:[a-zA-Z0-9-]+\.)+[a-zA-Z]{2,}\b"

domains = re.findall(pattern, text)
print(domains)  # ['google.com', 'mail.google.com', 'openai.org']
```
