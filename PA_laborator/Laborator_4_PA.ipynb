#### Exercițiul 2
Cifrul lui Cezar - folosind comprehensiune

````python
def shift_c(c, k):
    if 'a' <= c <= 'z':
        return chr((ord(c) - ord('a') + k) % 26 + ord('a'))
    else: 
        return c

k = 3
rez = [shift_c(ch, k) for ch in input()]
print(rez)
````

#### Exercițiul 18
Ciurul lui Eratostene: generați toate numerele prime ≤ n.

````python
def ciur(n):
    if n < 2:
        return []
    is_prime = [True] * (n + 1)
    is_prime[0] = is_prime[1] = False
    p = 2
    while p ** 2 <= n:
        if is_prime[p]:
            for m in range(p ** 2, n + 1, p):
                is_prime[m] = False
        p += 1
    return [i for i in range(2, n + 1) if  is_prime[i] == True]

print(ciur(30))    
````

#### Exercițiul 19
Numărul de perechi disjuncte de elemente egale dintr-un vector cu n<100 elemente de maxim 2 cifre. [Vector de frecvențe]
🔗 https://www.pbinfo.ro/probleme/2702/perechisosete

````python
def perechi(v):
    freq = [0] * 100
    for element in v:
        freq[element] += 1
    return sum(f // 2 for f in freq)


perechi([10, 10, 5, 5, 5, 99, 0, 0, 1])
````
