## 1) Varful Muntelui (Mountain Array Peak)

Se da un vector $a=(a_1,\ldots,a_n)$ de tip **munte** (exista un indice $i$ astfel incat $a_1 < a_2 < \ldots < a_i > a_{i+1} > \ldots > a_n$; $a_i$ se numeste **varful muntelui**).
Propuneti un algoritm **O(log n)** care determina varful muntelui (fara a considera citirea vectorului).

**Exemplu:**
| `date.in` | `date.out` |
|:---:|:---:|
| `5` | `11` |
| `4 8 10 11 5` | |

**Sugestie de algoritm $O(\log n)$ (Cautare Binara Modificata):**

1.  Se foloseste o abordare de tip **cautare binara** pe intervalul `[stanga, dreapta]`.
2.  La fiecare pas se calculeaza elementul de la mijloc, `mij = (stanga + dreapta) // 2`.
3.  Se compara $a_{mij}$ cu vecinul sau din dreapta, $a_{mij+1}$:
    * Daca $a_{mij} < a_{mij+1}$, inseamna ca ne aflam pe portiunea **crescatoare** a muntelui. Varful trebuie sa fie *in dreapta* (inclusiv $a_{mij+1}$). Se muta `stanga = mij + 1`.
    * Daca $a_{mij} > a_{mij+1}$, inseamna ca ne aflam pe portiunea **descrescatoare** a muntelui sau suntem *chiar la varf*. Varful poate fi $a_{mij}$ sau un element *din stanga* lui. Se muta `dreapta = mij`.
4.  Bucla se termina cand `stanga == dreapta`. Acesta va fi indicele varfului.

## 2) Numarul de Aparitii intr-o Lista Sortata

Scrieti o functie `nr_aparitii` cu complexitate **O(log n)** care primeste ca parametru o lista de numere intregi **ordonata crescator** si un numar $x$ si returneaza numarul de aparitii al valorii $x$ in lista.

**Exemplu:** `nr_aparitii( [1, 1, 2, 2, 2, 2, 6, 9, 9, 20], 2)` va returna $4$.

**Sugestie de algoritm $O(\log n)$:**

1.  Folositi **cautarea binara** de doua ori:
2.  Cautati `prima aparitie` a lui $x$ (cel mai mic indice $i$ pentru care $a_i = x$).
3.  Cautati `ultima aparitie` a lui $x$ (cel mai mare indice $j$ pentru care $a_j = x$).
4.  Numarul de aparitii este dat de formula: **$j - i + 1$**.
5.  Daca $x$ nu este gasit, rezultatul este $0$.

## 3) Mediana a Doi Vectori Sortati

Se dau doi vectori $a$ si $b$ de lungime $n$ (ambii), cu elementele **ordonate crescator**.
Propuneti un algoritm cat mai eficient pentru a determina **mediana** vectorului obtinut prin interclasarea celor doi vectori. **Complexitate tinta: $O(\log n)$**.

**Definitie:** Mediana unui vector sortat = elementul din mijlocul vectorului sau media aritmetica a celor doua elemente din mijloc (daca lungimea este para).
Vectorul interclasat are lungimea totala $2n$. Mediana este elementul de la pozitia $k=n$ sau $k=n-1$ si $k=n$.

**Exemplu:**
$n = 5$
$a$: `1 12 15 16 38`
$b$: `2 13 17 30 45`
Vector interclasat: `1, 2, 12, 13, 15, 16, 17, 30, 38, 45`
Lungime $2n=10$. Medianele sunt $15$ (pozitia 5) si $16$ (pozitia 6). Mediana finala ar fi $(15+16)/2 = 15.5$.

**Sugestie de algoritm $O(\log n)$ (Divide et Impera):**

* Se cauta elementul care ar fi pe pozitia $k=n$ (sau $k=n$ si $k=n+1$) in vectorul interclasat.
* Ideea principala este de a gasi o partitie a celor doi vectori, $(A_{st}, A_{dr})$ si $(B_{st}, B_{dr})$, astfel incat $|A_{st}| + |B_{st}| = k-1$ si $\max(\text{ultimul } A_{st}, \text{ultimul } B_{st}) \le \min(\text{primul } A_{dr}, \text{primul } B_{dr})$.
* Se foloseste o cautare binara pe intervalul de indici al primului vector ($a$) pentru a gasi split-ul optim.
* **Alternative simplificata:** Se poate folosi varianta de Divide et Impera care elimina recursiv jumatatile mai mici/mari care nu pot contine mediana.
    * Se compara medianele $m_a$ si $m_b$ ale subvectorilor curent.
    * Daca $m_a < m_b$, mediana globala nu poate fi in prima jumatate a lui $b$ sau in a doua jumatate a lui $a$.
    * Se continua recursiv pe subvectorii ramasi, reducand dimensiunea la jumatate la fiecare pas.

## 4) Perechi cu Suma Egala cu S

Se citeste o lista de numere naturale **sortata strict crescator** si un numar natural $S$.
Sa se afiseze toate **perechile distincte** formate din **valori distincte** din lista data cu proprietatea ca suma lor este egala cu $S$.

**Exemplu:**
Pentru lista `[1, 5, 8, 12, 15]` si $S = 16$, trebuie afisate perechile `(1, 15)` si `(8, 12)`.

**Sugestie de algoritm (Doua Puncte/Pointeri):**

* Desi nu este strict *Divide et Impera*, cel mai eficient algoritm este cu **doi pointeri** (sau "punctatori") care pornesc de la capetele listei:
    * Un pointer `stanga` (incepe la indicele 0).
    * Un pointer `dreapta` (incepe la indicele $n-1$).
* Se repeta cat timp `stanga < dreapta`:
    * Se calculeaza `suma = lista[stanga] + lista[dreapta]`.
    * Daca `suma == S`, se afiseaza perechea, si se muta ambii pointeri: `stanga++` si `dreapta--`.
    * Daca `suma < S`, pentru a creste suma, se muta `stanga++`.
    * Daca `suma > S`, pentru a micsora suma, se muta `dreapta--`.
* Complexitatea este **$O(n)$**, deoarece fiecare pointer parcurge lista o singura data.

## 5) Problema Debitarii (Max Area Rect)

Se considera o placa de tabla de forma dreptunghiulara avand coltul stanga-jos in punctul $(x_{min}, y_{min})$ si coltul dreapta-sus in punctul $(x_{max}, y_{max})$. Placa are pe suprafata sa $n$ gauri cu coordonate numere intregi.
Stiind ca sunt permise doar taieturi orizontale sau verticale **complete**, se cere sa se decupeze din placa o bucata de **arie maxima fara gauri**.

**Exemplu:**
Placa $(2, 1) - (8, 5)$. Gauri: $(3, 2), (4, 4), (5, 3), (7, 4)$.
`placa.out`: Dreptunghiul $(3, 1) - (8, 3)$, Aria maxima: $10$.

**Sugestie de algoritm (Divide et Impera bidimensional):**

* Aceasta problema este adesea rezolvata cu tehnici de **programare dinamica** sau **algoritmi geometrici $O(n^2)$/$O(n^3)$** (pentru gasirea celui mai mare dreptunghi in matrice cu restrictii).
* Totusi, daca se insista pe o abordare de tip **Divide et Impera**:
    1.  Se considera dreptunghiul actual (initial, toata placa).
    2.  Se alege o gaura mediana (de exemplu, cu coordonata $x$ sau $y$ mediana).
    3.  Aceasta gaura imparte dreptunghiul in maxim 4 sub-dreptunghiuri mai mici (sau 2, printr-o taietura completa pe verticala/orizontala gaurii).
    4.  Aria maxima este $\max(\text{Aria Maxima Recursiva pe Sub-dreptunghiuri})$.
* **Implementarea eficienta $O(n \log n)$** necesita o structura de date avansata (cum ar fi un **Quadtree/Kd-tree** sau o parcurgere cu **Sweep Line**), care ar iesi din scopul unui laborator standard. Pentru un laborator, o solutie recursiva simpla, dar mai putin eficienta (sau o solutie $O(n^2)$) poate fi acceptata, deși nu respecta cerinta stricta de $O(\log n)$ (care este nerealista pentru aceasta problema). O solutie de tip **Divide et Impera** care sorteaza gaurile pe axa $x$ si apoi pe axa $y$ si le imparte recursiv este o incercare de a aplica metoda, dar analiza complexitatii devine complicata.
